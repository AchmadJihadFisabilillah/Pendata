```python
import numpy as np
import pandas as pd

def wknn_imputer(data, target_idx, features, target_col, k=3):
    # Pisahkan data yang lengkap dan data yang hilang
    complete_data = data.drop(target_idx)
    target_row = data.loc[target_idx]
    
    distances = []
    for idx, row in complete_data.iterrows():
        # Hitung Euclidean Distance
        dist = np.sqrt(np.sum((row[features] - target_row[features])**2))
        distances.append((dist, row[target_col]))
    
    # Urutkan berdasarkan jarak terkecil
    distances.sort(key=lambda x: x[0])
    neighbors = distances[:k]
    
    # Hitung Weighted Average
    weighted_sum = 0
    total_weight = 0
    for dist, val in neighbors:
        weight = 1 / (dist + 1e-5)**2 # Tambah epsilon kecil agar tidak div by zero
        weighted_sum += weight * val
        total_weight += weight
        
    return weighted_sum / total_weight

# Data dari gambar (Normalized)
df = pd.DataFrame({
    'IPK': [0, 0.5, 1, 0, 0.5, 1, 0],
    'PO': [0, 0.5, 0, 0, 0.5, 1, 0.5],
    'JML': [0, 1, 0, 1, 0, 1, np.nan]
})

pred_jml = wknn_imputer(df, 6, ['IPK', 'PO'], 'JML', k=3)
print(f"Prediksi JML untuk baris 7: {pred_jml:.2f}")
```
